#### I want to make a dataframe which will have a series of car collections.

```python
import numpy as np
import pandas as pd
from io import StringIO
```

```python
c1=["0007","Aston Martin","DB5",233,282]
c2=["2207","Mercedes Benz","Gullwing",260,215]
c3=["1001","Ford Mustang","Boss 429",200,375]
c4=["8888","Porsche","918 Spyder",345,887]
c5=["2026","Ford Mustang","Fastback",193,271]
c6=["9247","Porsche","Macan",230,402]
c7=["7145","Bentley","Continental GT",335,659]
c8=["4997","Land Rover","Defender",240,296]
car_ar=np.array([c1,c2,c3,c4,c5,c6,c7,c8])
```
