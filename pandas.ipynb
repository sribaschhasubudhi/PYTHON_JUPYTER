```python
import numpy as np
import pandas as pd

# 1. Create fake sensor temperatures using NumPy (10 readings)
raw_temperatures = np.array([23.5, 24.1, 31.8, 22.9, 35.2, 26.4, 29.8, 33.1, 21.4, 30.5])

# 2. Convert that raw data into a clean Pandas DataFrame (Spreadsheet)
df = pd.DataFrame({
    'Sensor_ID': range(1, 11),
    'Temperature_C': raw_temperatures
})

# 3. Filter data: Find readings where the sensor is overheating (> 30°C)
overheating_sensors = df[df['Temperature_C'] > 30]

print("--- All Sensor Data ---")
print(df)
print("\n--- Overheating Sensors (> 30°C) ---")
print(overheating_sensors)
```
