```python
import numpy as np
import pandas as pd
from io import StringIO
```

```python
xml_1="""<catalog>
    <book>
        <id>1</id>
        <author>Cal Newport</author>
        <title>Deep Work</title>
    </book>
</catalog>"""
pd.read_xml(StringIO(xml_1))
```

```python
xml_2="""<catalog>
    <book id="1" author="Cal Newport" title="Deep Work" />
</catalog>"""
pd.read_xml(StringIO(xml_2),attrs_only=True)
```

```python
xml_3="""<catalog>
    <book id="1" status="published">
        <author>Cal Newport</author>
        <title>Deep Work</title>
    </book>
</catalog>"""
pd.read_xml(StringIO(xml_3))
```
