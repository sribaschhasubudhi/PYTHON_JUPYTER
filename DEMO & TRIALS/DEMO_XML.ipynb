~ First, we will create xml files and then we will use them.

```python
import numpy as np
import pandas as pd
from io import StringIO
```

```python
job_data={"ID":[17,59,35,29,40],
          "Name":["Jack","Anne","Peter","Daniel","Jovile"],
          "Profession":["Doctor","Mechanic","Manager","Engineer",
                        "Teacher"]}
df=pd.DataFrame(job_data)
df.to_xml("FILES_JUPYTER/jobs.xml",index=False)
```

```python
job_att=df.to_xml(index=False,attr_cols=["ID","Name","Profession"])
# This is not permanently saved
```

```python
print(job_att)
```

```python
pd.read_xml(StringIO(job_att),xpath="./*")      #DEFAULT
pd.read_xml(StringIO(job_att),xpath=".//row")   # NO CHANGE
pd.read_xml(StringIO(job_att))                  #DEFAULT
```

```python
pd.read_xml("FILES_JUPYTER/jobs.xml",elems_only=True)
# "FILES_JUPYTER/jobs.xml" doesn't have any value in container attributes.
# It is in element-centric format.
```

```python
pd.read_xml(StringIO(job_att),attrs_only=True)
# job_att does have any child elements in text tags.
# It is in attribute-centric format
```
