~ We will now create and read a pkl file by importing pickle

```python
import pickle
```

```python
dit={"C0":[12,64,31,60,80],
     "C1":[35,71,29,49,84],
     "C2":[30,64,91,53,64],
     "C3":[15,12,93,np.nan,85],
     "C4":[63,52,95,31,63]
    }
with open("FILES_JUPYTER/numbers.pkl","wb") as file:
    pickle.dump(dit,file)
print(type(file))
''' format of dump:- pickle.dump(data variable,file object)
in this case, data variable is dit and file object is file'''
```

```python
with open("FILES_JUPYTER/numbers.pkl","rb") as file:
    load_nos=pickle.load(file)
print(type(file))
'''format of load:- pickle.load(file object)'''
```

=> the type of 'file' in pickle.dump() case is 'bufferedwriter' but in case of pickle.load(), it is 'bufferedreader'.

```python
print(type(load_nos))
print(load_nos)
print(load_nos["C3"])
```
