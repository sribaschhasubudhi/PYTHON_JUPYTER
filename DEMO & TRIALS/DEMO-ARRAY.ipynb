## USING ARRAY ##

```python
import numpy as np
```

```python
t=(2,5,7,3)
l=[2,5,7,3]
d={'a':1,'b':3,'c':5}
```

```python
ar=np.array(l)
ar1=np.array(t)
ar2=np.array(d)
print(ar)
print(ar1)
print(ar2)
```

```python
type(ar)    #both type is of 1-dimension
type(ar2)
```
