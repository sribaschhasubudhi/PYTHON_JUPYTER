### CREATING A LOTTERY SYSTEM
##### In this system, participants have to buy tickets which costs 500$.
##### The one having the majority number of tickets, wins.
##### The  total money collected by selling the tickets will be divided in half.
##### The winner will be given half the amount from the collection.

```python
import random
```

```python
# Tickets
tickets=[]
#random.seed(73.8) gives 2 same no.of tickets when range is (1,9)
#random.seed(73.8)
for i in range(0,5):
    no_of=random.randint(1,9)
    tickets.append(no_of)
print(f"Tickets:- {tickets}")
total_tickets=sum(tickets)
print(f"Total number of tickets sold:-{total_tickets}")
total_collection=total_tickets*500
print(f"Total money collected:-{total_collection:,}")
```

```python
# {Name: Tickets}
name=["Alice","Jack","Jennie","Stephan","Sarah"]
random.shuffle(name)
random.shuffle(tickets)
participants={}
for i in range(0,5):
    participants[name[i]]=tickets[i]
print(f"Lottery slot:-{participants}")
```

#### Everything is random now. Those who buyed more or less tickets, doesn't matter. Anybody can win now.

```python
win=[]
random_tickets=participants.values()
high=max(random_tickets)
for name,tickets in participants.items():
    if tickets==high:
        win.append(name)
print(f"Potential winners:-{win}")

'''This one is a very important step'''
```

```python
# Winner
if len(win)>1:
    winner=random.choice(win)
    print(f"The winner is {winner}")
    print(f"Woohoo! {winner} gets a cash reward of ${total_collection/2:,} dollars")
else:
    print(f"The winner is {win[0]}")
    print(f"Woohoo! {win[0]} gets a cash reward of ${total_collection/2:,} dollars")
```
