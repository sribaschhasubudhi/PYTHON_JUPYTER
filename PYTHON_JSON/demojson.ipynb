~ Creating a attendance sheet of a school having 3 sections:A,B,C   
~ Each section will have some students and they will be marked **present or absent**.

```python
import pandas as pd
import numpy as np
from io import StringIO
```

```python
#First we need to create a dataframe.
'''Blueprint:-
    sections:- A,B,C
        A:-John, Adam, Rick
        B:-Bill, Jane, Kenny
        C:-Walton, Isabel, Lili'''
```

```python
# This one is for dataframe
st_dict={"A":["John","Adam","Rick"],
         "B":["Bill","Jane","Kenny"],
         "C":["Walton","Isabel","Lili"]
        }
```

~~ While using a dictionary to create a dataframe, the keys become the column names while   the values are the elements of the respective key-columns.

```python
stdf=pd.DataFrame(st_dict)
stdf
```

```python
stdf.to_json("FILES_JUPYTER/STUDENTS.JSON",indent=4)
stdf.to_json()
```

```python
att_dt=[{"Section":"A",
        "students":["John","Adam","Rick"],
        "attendance":
                [{"John":"Present"},
                 {"Adam":"Absent"},
                 {"Rick":"Absent"}
                 ]},
        {"Section":"B",
        "students":["Bill","Jane","Kenny"],
        "attendance":
                [{"Bill":"Present"},
                 {"Jane":"Absent"},
                 {"Kenny":"Present"}
                 ]},
        {"Section":"C",
        "students":["Walton","Isabel","Lili"],
        "attendance":
                [{"Walton":"Present"},
                 {"Isabel":"Absent"},
                 {"Lili":"Present"}
                 ]}
        ]
# att_dt is in json format
```

```python
att_df=pd.DataFrame(att_dt)
att_df
```

```python
att_j=att_df.to_json("FILES_JUPYTER/attendace.json",indent=4)
```

```python
att_df.to_json()
```

#### Let's create another json data on the book:- "Deep work"

```python
'''Blueprint:-
    book:- Deep Work
    author:- Cal Newport
    details:-
        pages:- 304,
        cover type:- Paperback,
        ISBN:- 9781455586691'''
```

```python
book_dat={"Book":"Deep Work",
          "Author":"Cal Newport",
          "Details":[{"pages":304,
                      "cover type":"Papertype",
                      "ISBN":9781455586691}
                    ]
        }
book_df=pd.DataFrame(book_dat)
book_df
```

=> Now we have 3 json datas in this demojson.pynb.   
=> st_dict and att_dt had been saved in form of json files named
    "attendance.json" and "STUDENTS.JSON" respectively.  
=> book_dat is present in form of a dataframe and in json format. 

```python
book_df.to_json("FILES_JUPYTER/Deep Work.json",indent=4)
```
