## groupby():-

```python
from itertools import groupby
```

```python
data1=[("A",3),("A",3),("A",6),("A",10),("B",34),("B",18),("B",84),("C",19),("C",20)]
data2=[("A",3),("A",6),("A",10),("a",16),("B",34),("B",18),("B",84),("C",19),("C",20),("D",)]
```

```python
for key,group in groupby(data1,key=None):
    print(key,list(group))
print("-"*70)
for key,group in groupby(data2):
    print(key,list(group))
```

```python
for key,group in groupby(data1,key=lambda x:x[0]):
    print(key,next(group))                      # Using next()
print("÷"*5)
for key,group in groupby(data1,key=lambda x:x[0]):
    print(list(group))                          # Using list()                    
```

```python
for key,group in groupby(data2,key=lambda x:x[0]):
    print(key,list(group))
    print(group)
```

```python
vegan=["apple","banana","Cucumber","carrot","dates","eggplant"]
vegan_small=[]
for i in vegan:
    vegan_small.append(i.capitalize())
print(vegan_small)

grp_veg=groupby(sorted(vegan_small),lambda x:x[0])      # Same output with key="lambda..."
for key, grp in grp_veg:
    print(key,list(grp))   
```

```python
data2=[("A",3),("A",6),("A",10),("a",16),("B",34),("B",18),("B",84),("C",19),("C",20),("D",)]
for key,group in groupby(sorted(data2,reverse=True),key=lambda x:x[0]):
    print(key,list(group))
# "reverse=True" to show it in ascending order
```

## islice():-

```python
from itertools import islice
```

```python
num1=[12,5,9,23,60,57,20,4,18]
it_num1=iter(num1)
print(it_num1)
for i in it_num1:
    print(i)
```

```python
print(list(islice(it_num1,5)))     # The above iterator ate every element. So, it prints a empty list.
it_num1=iter(num1)                 # New iterator formed                
print(list(islice(it_num1,5)))
```

```python
num2=[12,5,9,23,60,57,20,4,18,92,50,16,36,51,91,76,82,61,26,8,7,3,865,532,75,1,98,7]
print(f"{len(num2)} elements are present in num2")
print(islice(num1,5))               # Shows islice as an object
print(list(islice(num1,5)))         # Here, 5 is the stop value. Same as num1[0:5]
it_num2=iter(num2)
print(next(it_num2))
print(list(islice(it_num2,6)))      # Stops at 6th index value
print(list(islice(it_num2,6,7)))    # Start value is 6 and stop value is 7
```

```python
go2=count(2,step=3)                  # Start value is 2 and step value is 3
print(list(islice(go2,8)))           # Prints upto 8th index value
'''print(next(islice(go2,0)))'''     # Shows StopIteration
print(next(go2))
print(list(islice(go2,8)))           # Prints the next values
```

## accumulate():-

```python
from itertools import accumulate
import operator
```

```python
num3=[1,2,3,4,5]
print(list(accumulate(num3)))
print(accumulate(num3))                 # Prints it as object
go3=count(1)
print(list(accumulate(islice(go3,8))))
print(list(accumulate(islice(go3,8))))  # Prints the next elements due to iterator
```

## compress():-

```python
from itertools import compress
```

```python
build=["Cement","Pebbles","Sand","Shovel","Bricks","Rod"]
check1=[1,0,0,1,0,1]
check2=[1,0,0,0,1]                              # 5 checks provided
check3=[0,0,0,0,000,0]
print(compress(build,check1))                   # Shows the compress object
print(list(compress(build,selectors=check1)))
print(list(compress(sorted(build),check1)))
print(list(compress(build,selectors=check2)))   # Unique!
print(next(compress(build,check1)))
print(list(compress(build,check3)))             # Empty list 
```
